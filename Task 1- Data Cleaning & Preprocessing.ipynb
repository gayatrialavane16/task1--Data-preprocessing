```python
import pandas as pd
df = pd.read_csv("D:\\Intership\\Elevate_Lab\\Titanic-Dataset.csv")   # adjust path if it's not in the same folder
df.head()
```

```python
df.info()
df.isnull().sum()
df.describe()
```

```python
# Age - fill with median (numeric, skewed)
df['Age'].fillna(df['Age'].median(), inplace=True)

# Embarked - fill with mode (categorical, few missing)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# Cabin - too many missing (~77%), drop the column
df.drop(columns=['Cabin'], inplace=True)

# verify
df.isnull().sum()
```

```python
# Sex - binary, use label encoding
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# Embarked - more than 2 categories, use one-hot encoding
df = pd.get_dummies(df, columns=['Embarked'], drop_first=True)

df.head()
```

```python
df.drop(columns=['PassengerId', 'Name', 'Ticket'], inplace=True, errors='ignore')

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])

df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# visualize outliers
plt.figure(figsize=(10,4))
sns.boxplot(data=df[['Age', 'Fare']])
plt.show()

# remove outliers using IQR method (on Fare, since it usually has more extreme outliers)
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df['Fare'] >= lower) & (df['Fare'] <= upper)]

print(df.shape)
```
